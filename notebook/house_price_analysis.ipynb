```python
import pandas as pd

df = pd.read_csv("../data/cleaned_house_prices.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
print(df.head())
```

```python
correlation = df[["area_sqft", "bedrooms", "bathrooms", "parking", "price"]].corr()

print(correlation)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8, 6))

sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Heatmap")
plt.show()
```

```python
bedroom_price = df.groupby("bedrooms")["price"].mean()

print("Average price by bedrooms:")
print(bedroom_price)
```

```python
bathroom_price = df.groupby("bathrooms")["price"].mean()

print("Average price by bathrooms:")
print(bathroom_price)
```

```python
parking_price = df.groupby("parking")["price"].mean()

print("Average price by parking:")
print(parking_price)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.scatter(df["area_sqft"], df["price"])

plt.xlabel("Area (sqft)")
plt.ylabel("Price (₹)")
plt.title("Area vs Price")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(x="location", y="price", data=df)

plt.xlabel("Location")
plt.ylabel("Price (₹)")
plt.title("House Price Distribution by Location")

plt.show()
```

```python
import pandas as pd

df = pd.read_csv("../data/cleaned_house_prices.csv")

print("Dataset loaded!")
print("Shape:", df.shape)
print(df.head())
```

```python
price_sqft_by_location = df.groupby("location")["price_per_sqft"].mean()

print("Average price per square foot by location:")
print(price_sqft_by_location)
```

```python
most_expensive_sqft = df.loc[df["price_per_sqft"].idxmax()]

print("House with the highest price per square foot:")
print(most_expensive_sqft)
```

```python
average_area_by_location = df.groupby("location")["area_sqft"].mean()

print("Average house area by location:")
print(average_area_by_location)
```

```python
average_bedrooms_by_location = df.groupby("location")["bedrooms"].mean()

print("Average bedrooms by location:")
print(average_bedrooms_by_location)
```

```python
average_bathrooms_by_location = df.groupby("location")["bathrooms"].mean()

print("Average bathrooms by location:")
print(average_bathrooms_by_location)
```

```python
parking_by_location = df.groupby("location")["parking"].mean()

print("Average parking availability by location:")
print(parking_by_location)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

average_area_by_location.plot(kind="bar")

plt.xlabel("Location")
plt.ylabel("Average Area (sqft)")
plt.title("Average House Area by Location")
plt.xticks(rotation=0)

plt.show()
```

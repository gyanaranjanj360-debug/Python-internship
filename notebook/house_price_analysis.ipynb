```python
import pandas as pd

df = pd.read_csv("../data/house_price.csv")

df.head()
```

```python
print("Number of rows and columns:", df.shape)
```

```python
print("Column Names:")
print(df.columns)
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
df.describe()
```

```python
print("Data Types:")
print(df.dtypes)
```

```python
print("Locations in the dataset:")
print(df["location"].unique())
```

```python
print("Number of houses in each location:")
print(df["location"].value_counts())
```

```python
print("Average house price by location:")
print(df.groupby("location")["price"].mean())
```

```python
df["price_per_sqft"] = df["price"] / df["area_sqft"]

print("Price per square foot:")
print(df[["location", "area_sqft", "price", "price_per_sqft"]])
```

```python
df.to_csv("../data/cleaned_house_prices.csv", index=False)

print("Cleaned dataset saved successfully!")
```

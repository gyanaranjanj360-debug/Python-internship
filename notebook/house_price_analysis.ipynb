```python
import pandas as pd

df = pd.read_csv("../data/cleaned_house_prices.csv")

print("Dataset loaded successfully!")
```

```python
print("Average house price:")
print(df["price"].mean())
```

```python
print("Cheapest house:")
print(df.loc[df["price"].idxmin()])
```

```python
print("Most expensive house:")
print(df.loc[df["price"].idxmax()])
```

```python
print("Average price by location:")

print(
    df.groupby("location")["price"]
    .mean()
    .sort_values(ascending=False)
)
```

```python
print("Average price per square foot by location:")

print(
    df.groupby("location")["price_per_sqft"]
    .mean()
    .sort_values(ascending=False)
)
```

```python
import matplotlib.pyplot as plt

print("Matplotlib imported successfully!")
```

```python
avg_price = df.groupby("location")["price"].mean()

avg_price.plot(kind="bar")

plt.title("Average House Price by Location")
plt.xlabel("Location")
plt.ylabel("Average Price (₹)")
plt.xticks(rotation=45)

plt.show()
```

```python
plt.scatter(df["area_sqft"], df["price"])

plt.title("House Area vs Price")
plt.xlabel("Area (sqft)")
plt.ylabel("Price (₹)")

plt.show()
```

```python
bedroom_price = df.groupby("bedrooms")["price"].mean()

bedroom_price.plot(kind="bar")

plt.title("Average House Price by Bedrooms")
plt.xlabel("Number of Bedrooms")
plt.ylabel("Average Price (₹)")

plt.show()
```

```python
price_sqft = df.groupby("location")["price_per_sqft"].mean()

price_sqft.plot(kind="bar")

plt.title("Average Price per Sqft by Location")
plt.xlabel("Location")
plt.ylabel("Price per Sqft (₹)")
plt.xticks(rotation=45)

plt.show()
```

```python
print("EDA SUMMARY")
print("--------------------")

print("Total houses:", len(df))
print("Average price: ₹", round(df["price"].mean(), 2))
print("Minimum price: ₹", df["price"].min())
print("Maximum price: ₹", df["price"].max())

print("\nAverage price by location:")
print(df.groupby("location")["price"].mean())
```

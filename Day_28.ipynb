```python
# Day 28: Location-Based House Price Prediction

print("Day 28 - Location-Based Price Prediction")
print("Project: Smart House Price Predictor")
```

```python
import pandas as pd

data = {
    "area_sqft": [800, 1000, 1200, 1500, 1800, 2000,
                  1000, 1300, 1600, 2000, 900, 1400],
    "bedrooms": [2, 2, 3, 3, 3, 4, 2, 3, 3, 4, 2, 3],
    "bathrooms": [1, 2, 2, 2, 3, 3, 1, 2, 2, 3, 1, 2],
    "parking": [0, 1, 1, 1, 1, 2, 0, 1, 1, 2, 0, 1],
    "location": [
        "Cuttack", "Cuttack", "Bhubaneswar", "Bhubaneswar",
        "Bhubaneswar", "Bhubaneswar", "Puri", "Puri",
        "Puri", "Puri", "Berhampur", "Berhampur"
    ],
    "price": [2000000, 2500000, 3500000, 4500000, 5200000,
              6500000, 2400000, 3200000, 4000000, 5000000,
              1800000, 3000000]
}

df = pd.DataFrame(data)
display(df)
```

```python
from sklearn.tree import DecisionTreeRegressor

X = pd.get_dummies(
    df[["area_sqft", "bedrooms", "bathrooms", "parking", "location"]],
    columns=["location"],
    dtype=int
)

y = df["price"]

model = DecisionTreeRegressor(max_depth=4, random_state=42)
model.fit(X, y)

print("Location-based model trained successfully!")
print("Model features:", list(X.columns))
```

```python
def predict_location_price(area, bedrooms, bathrooms, parking, location):
    if area <= 0 or bedrooms < 1 or bathrooms < 1 or parking < 0:
        raise ValueError("Please enter valid house details.")

    if location not in df["location"].unique():
        raise ValueError("Location not found in the dataset.")

    house = pd.DataFrame([{
        "area_sqft": area,
        "bedrooms": bedrooms,
        "bathrooms": bathrooms,
        "parking": parking,
        "location": location
    }])

    house_encoded = pd.get_dummies(
        house, columns=["location"], dtype=int
    )

    house_encoded = house_encoded.reindex(
        columns=X.columns, fill_value=0
    )

    return model.predict(house_encoded)[0]

print("Prediction function created successfully!")
```

```python
predicted_price = predict_location_price(
    area=1500,
    bedrooms=3,
    bathrooms=2,
    parking=1,
    location="Bhubaneswar"
)

print("Location: Bhubaneswar")
print("Predicted price: ₹", round(predicted_price, 2))
print("Price in lakhs: ₹", round(predicted_price / 100000, 2), "lakh")
```

```python
location_results = []

for location in sorted(df["location"].unique()):
    price = predict_location_price(
        area=1500,
        bedrooms=3,
        bathrooms=2,
        parking=1,
        location=location
    )

    location_results.append({
        "Location": location,
        "Predicted Price": round(price, 2),
        "Price in Lakhs": round(price / 100000, 2)
    })

location_df = pd.DataFrame(location_results)
display(location_df)

location_df.to_csv("day28_location_predictions.csv", index=False)
print("Results saved successfully!")
```

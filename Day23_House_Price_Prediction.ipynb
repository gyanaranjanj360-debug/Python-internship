```python
import pandas as pd

data = {
    "area_sqft": [800, 1000, 1200, 1500, 1800, 2000, 2200, 2500],
    "bedrooms":  [2, 2, 3, 3, 3, 4, 4, 4],
    "bathrooms": [1, 2, 2, 2, 3, 3, 3, 4],
    "parking":   [0, 1, 1, 1, 1, 2, 2, 2],
    "price":     [2000000, 2500000, 3200000, 4000000,
                  4800000, 5500000, 6200000, 7500000]
}

df = pd.DataFrame(data)

print("Sample dataset created successfully!")
display(df)
```

```python
from sklearn.tree import DecisionTreeRegressor

X = df[["area_sqft", "bedrooms", "bathrooms", "parking"]]
y = df["price"]

demo_model = DecisionTreeRegressor(max_depth=3, random_state=42)
demo_model.fit(X, y)

print("Demonstration model trained successfully!")
```

```python
def predict_house_price(area, bedrooms, bathrooms, parking):
    house = pd.DataFrame([{
        "area_sqft": area,
        "bedrooms": bedrooms,
        "bathrooms": bathrooms,
        "parking": parking
    }])

    predicted_price = demo_model.predict(house)[0]
    return predicted_price

print("Prediction function created successfully!")
```

```python
price = predict_house_price(
    area=1500,
    bedrooms=3,
    bathrooms=2,
    parking=1
)

print("Estimated House Price: ₹", round(price, 2))
print("Estimated Price in Lakhs: ₹", round(price / 100000, 2), "lakh")
```

```python
def predict_house_price(area, bedrooms, bathrooms, parking):
    if area <= 0:
        raise ValueError("Area must be greater than zero.")

    if bedrooms < 1 or bathrooms < 1 or parking < 0:
        raise ValueError("Please enter valid house details.")

    house = pd.DataFrame([{
        "area_sqft": area,
        "bedrooms": bedrooms,
        "bathrooms": bathrooms,
        "parking": parking
    }])

    return demo_model.predict(house)[0]

print("Prediction function with validation is ready!")
```

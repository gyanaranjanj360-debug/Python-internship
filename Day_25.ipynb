```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "area_sqft": [800, 1000, 1200, 1500, 1800, 2000, 2200, 2500],
    "bedrooms": [2, 2, 3, 3, 3, 4, 4, 4],
    "bathrooms": [1, 2, 2, 2, 3, 3, 3, 4],
    "parking": [0, 1, 1, 1, 1, 2, 2, 2],
    "actual_price": [
        2000000, 2500000, 3200000, 4000000,
        4800000, 5500000, 6200000, 7500000
    ]
}

df = pd.DataFrame(data)
display(df)
```

```python
from sklearn.tree import DecisionTreeRegressor

X = df[["area_sqft", "bedrooms", "bathrooms", "parking"]]
y = df["actual_price"]

model = DecisionTreeRegressor(max_depth=3, random_state=42)
model.fit(X, y)

df["predicted_price"] = model.predict(X)

display(df[["actual_price", "predicted_price"]])
```

```python
df["prediction_error"] = (
    df["actual_price"] - df["predicted_price"]
)

df["absolute_error"] = df["prediction_error"].abs()

print("Mean Absolute Error: ₹", round(df["absolute_error"].mean(), 2))
display(df[["actual_price", "predicted_price", "prediction_error"]])
```

```python
plt.figure(figsize=(8, 5))

plt.plot(df["actual_price"], marker="o", label="Actual Price")
plt.plot(df["predicted_price"], marker="s", label="Predicted Price")

plt.title("Actual vs Predicted House Prices")
plt.xlabel("House Number")
plt.ylabel("Price (₹)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import pandas as pd

data = {
    "area_sqft": [800, 1000, 1200, 1500, 1800, 2000, 2200, 2500,
                  900, 1300, 1600, 2100],
    "bedrooms":  [2, 2, 3, 3, 3, 4, 4, 4, 2, 3, 3, 4],
    "bathrooms": [1, 2, 2, 2, 3, 3, 3, 4, 1, 2, 2, 3],
    "parking":   [0, 1, 1, 1, 1, 2, 2, 2, 0, 1, 1, 2],
    "price":     [2000000, 2500000, 3200000, 4000000, 4800000, 5500000,
                  6200000, 7500000, 2200000, 3300000, 4200000, 6000000]
}

df = pd.DataFrame(data)
display(df.head())
print("Sample dataset created!")
```

```python
from sklearn.tree import DecisionTreeRegressor

X = df[["area_sqft", "bedrooms", "bathrooms", "parking"]]
y = df["price"]

model = DecisionTreeRegressor(max_depth=3, random_state=42)
model.fit(X, y)

print("Model trained successfully!")
```

```python
def predict_house_price(area, bedrooms, bathrooms, parking):
    if area <= 0:
        raise ValueError("Area must be greater than zero.")
    if bedrooms < 1 or bathrooms < 1 or parking < 0:
        raise ValueError("Enter valid house details.")

    house = pd.DataFrame([{
        "area_sqft": area,
        "bedrooms": bedrooms,
        "bathrooms": bathrooms,
        "parking": parking
    }])

    return model.predict(house)[0]

print("Prediction function is ready!")
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output

area_input = widgets.IntSlider(
    value=1500, min=500, max=3000, step=100,
    description="Area (sq ft)"
)

bedrooms_input = widgets.IntSlider(
    value=3, min=1, max=6, description="Bedrooms"
)

bathrooms_input = widgets.IntSlider(
    value=2, min=1, max=5, description="Bathrooms"
)

parking_input = widgets.IntSlider(
    value=1, min=0, max=3, description="Parking"
)

predict_button = widgets.Button(
    description="Predict Price",
    button_style="success"
)

output = widgets.Output()

def on_predict_clicked(button):
    with output:
        clear_output(wait=True)
        price = predict_house_price(
            area_input.value,
            bedrooms_input.value,
            bathrooms_input.value,
            parking_input.value
        )
        print("Estimated house price: ₹", f"{price:,.0f}")
        print(f"Estimated price: ₹{price / 100000:.2f} lakh")

predict_button.on_click(on_predict_clicked)

display(
    area_input,
    bedrooms_input,
    bathrooms_input,
    parking_input,
    predict_button,
    output
)
```

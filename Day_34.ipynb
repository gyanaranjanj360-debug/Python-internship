```python
import pandas as pd
import numpy as np

data = {
    "area_sqft": [800, 1000, 1200, 1500, 1800, 2000, 2200, 2500],
    "bedrooms": [2, 2, 3, 3, 3, 4, 4, 4],
    "bathrooms": [1, 2, 2, 2, 3, 3, 3, 4],
    "parking": [0, 1, 1, 1, 1, 2, 2, 2],
    "location": [
        "Cuttack", "Cuttack", "Bhubaneswar", "Bhubaneswar",
        "Puri", "Puri", "Berhampur", "Berhampur"
    ],
    "price": [
        2000000, 2500000, 3200000, 4000000,
        4800000, 5500000, 6200000, 7500000
    ]
}

df = pd.DataFrame(data)

# Add sample missing values for preprocessing practice
df.loc[2, "bathrooms"] = np.nan
df.loc[5, "parking"] = np.nan

display(df)
```

```python
print("Missing values before cleaning:")
print(df.isnull().sum())

df["bathrooms"] = df["bathrooms"].fillna(
    df["bathrooms"].median()
)

df["parking"] = df["parking"].fillna(
    df["parking"].median()
)

print("\nMissing values after cleaning:")
print(df.isnull().sum())
```

```python
df["price_per_sqft"] = df["price"] / df["area_sqft"]

display(df[["area_sqft", "location", "price", "price_per_sqft"]])
print("Price per square foot feature created!")
```

```python
df_encoded = pd.get_dummies(
    df,
    columns=["location"],
    dtype=int
)

display(df_encoded.head())
print("Location encoding completed!")
```

```python
df_encoded.to_csv(
    "day34_processed_house_prices.csv",
    index=False
)

print("Processed dataset saved successfully!")
```

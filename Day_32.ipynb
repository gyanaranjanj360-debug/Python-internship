```python
# Day 32: Save and Load a Machine Learning Model

print("Day 32 - Save and Load ML Model")
print("Project: Smart House Price Predictor")
```

```python
import pandas as pd

data = {
    "area_sqft": [800, 1000, 1200, 1500, 1800, 2000,
                  2200, 2500, 900, 1300, 1600, 2100],
    "bedrooms":  [2, 2, 3, 3, 3, 4, 4, 4, 2, 3, 3, 4],
    "bathrooms": [1, 2, 2, 2, 3, 3, 3, 4, 1, 2, 2, 3],
    "parking":   [0, 1, 1, 1, 1, 2, 2, 2, 0, 1, 1, 2],
    "price":     [2000000, 2500000, 3200000, 4000000, 4800000,
                  5500000, 6200000, 7500000, 2200000, 3300000,
                  4200000, 6000000]
}

df = pd.DataFrame(data)
display(df.head())
```

```python
from sklearn.tree import DecisionTreeRegressor

X = df[["area_sqft", "bedrooms", "bathrooms", "parking"]]
y = df["price"]

model = DecisionTreeRegressor(max_depth=3, random_state=42)
model.fit(X, y)

print("Model trained successfully!")
```

```python
import joblib

joblib.dump(model, "day32_house_price_model.pkl")

print("Model saved as day32_house_price_model.pkl")
```

```python
loaded_model = joblib.load("day32_house_price_model.pkl")

print("Saved model loaded successfully!")
```

```python
new_house = pd.DataFrame([{
    "area_sqft": 1500,
    "bedrooms": 3,
    "bathrooms": 2,
    "parking": 1
}])

predicted_price = loaded_model.predict(new_house)[0]

print("Estimated house price: ₹", round(predicted_price, 2))
print("Estimated price in lakhs: ₹", round(predicted_price / 100000, 2), "lakh")
```

```python
original_prediction = model.predict(new_house)[0]
loaded_prediction = loaded_model.predict(new_house)[0]

print("Original model prediction: ₹", round(original_prediction, 2))
print("Loaded model prediction: ₹", round(loaded_prediction, 2))

assert original_prediction == loaded_prediction
print("Verification successful: both predictions match!")
```

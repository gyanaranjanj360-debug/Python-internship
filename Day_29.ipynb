```python
# Day 29: House Price Market Analysis

print("Day 29 - House Price Market Analysis")
print("Project: Smart House Price Predictor and Market Analysis")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "location": [
        "Bhubaneswar", "Bhubaneswar", "Bhubaneswar",
        "Cuttack", "Cuttack", "Cuttack",
        "Puri", "Puri", "Puri",
        "Berhampur", "Berhampur", "Berhampur"
    ],
    "area_sqft": [
        1200, 1500, 1800, 1000, 1300, 1600,
        900, 1400, 1700, 1000, 1400, 1600
    ],
    "price": [
        3500000, 4500000, 5200000,
        2500000, 3200000, 4000000,
        2400000, 3200000, 4200000,
        2200000, 3000000, 3600000
    ]
}

df = pd.DataFrame(data)
df["price_per_sqft"] = df["price"] / df["area_sqft"]

display(df)
print("Sample market dataset created!")
```

```python
market_summary = df.groupby("location").agg(
    average_price=("price", "mean"),
    average_price_per_sqft=("price_per_sqft", "mean"),
    number_of_houses=("price", "count")
).reset_index()

market_summary["average_price_lakhs"] = (
    market_summary["average_price"] / 100000
)

market_summary["average_price_per_sqft"] = (
    market_summary["average_price_per_sqft"].round(2)
)

display(market_summary.round(2))
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    market_summary["location"],
    market_summary["average_price_lakhs"]
)

plt.title("Average House Price by Location")
plt.xlabel("Location")
plt.ylabel("Average Price (₹ lakh)")
plt.grid(axis="y")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    market_summary["location"],
    market_summary["average_price_per_sqft"]
)

plt.title("Average Price per Square Foot")
plt.xlabel("Location")
plt.ylabel("Price per Sq. Ft. (₹)")
plt.grid(axis="y")
plt.show()
```

```python
market_summary.to_csv(
    "day29_house_price_market_summary.csv",
    index=False
)

print("Market summary saved successfully!")
```

```python
import pandas as pd
from sklearn.tree import DecisionTreeRegressor

data = {
    "area_sqft": [800, 1000, 1200, 1500, 1800, 2000, 2200, 2500],
    "bedrooms":  [2, 2, 3, 3, 3, 4, 4, 4],
    "bathrooms": [1, 2, 2, 2, 3, 3, 3, 4],
    "parking":   [0, 1, 1, 1, 1, 2, 2, 2],
    "price":     [2000000, 2500000, 3200000, 4000000,
                  4800000, 5500000, 6200000, 7500000]
}

df = pd.DataFrame(data)
display(df)
print("Dataset created successfully!")
```

```python
X = df[["area_sqft", "bedrooms", "bathrooms", "parking"]]
y = df["price"]

model = DecisionTreeRegressor(max_depth=3, random_state=42)
model.fit(X, y)

print("Model trained successfully!")
```

```python
def predict_house_price(area, bedrooms, bathrooms, parking):
    if area <= 0:
        raise ValueError("Area must be greater than zero.")
    if bedrooms < 1 or bathrooms < 1 or parking < 0:
        raise ValueError("Enter valid house details.")

    house = pd.DataFrame([{
        "area_sqft": area,
        "bedrooms": bedrooms,
        "bathrooms": bathrooms,
        "parking": parking
    }])

    return model.predict(house)[0]

print("Prediction function created successfully!")
```

```python
predicted_price = predict_house_price(
    area=1500,
    bedrooms=3,
    bathrooms=2,
    parking=1
)

print("Predicted house price: ₹", round(predicted_price, 2))
print("Predicted price in lakhs: ₹", round(predicted_price / 100000, 2), "lakh")
```

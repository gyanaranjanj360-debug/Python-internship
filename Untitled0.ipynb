```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "location": [
        "Bhubaneswar", "Bhubaneswar", "Bhubaneswar",
        "Cuttack", "Cuttack", "Cuttack",
        "Puri", "Puri", "Puri",
        "Berhampur", "Berhampur", "Berhampur"
    ],
    "area_sqft": [
        1200, 1500, 1800, 1000, 1300, 1600,
        900, 1400, 1700, 1000, 1400, 1600
    ],
    "bedrooms": [2, 3, 3, 2, 3, 3, 2, 3, 3, 2, 3, 3],
    "price": [
        3500000, 4500000, 5200000,
        2500000, 3200000, 4000000,
        2400000, 3200000, 4200000,
        2200000, 3000000, 3600000
    ]
}

df = pd.DataFrame(data)
df["price_per_sqft"] = df["price"] / df["area_sqft"]

display(df.head())
print("Dataset created successfully!")
```

```python
total_houses = len(df)
average_price = df["price"].mean()
highest_price = df["price"].max()
lowest_price = df["price"].min()

print("HOUSE PRICE DASHBOARD")
print("-" * 30)
print("Total houses:", total_houses)
print("Average price: ₹", round(average_price, 2))
print("Highest price: ₹", round(highest_price, 2))
print("Lowest price: ₹", round(lowest_price, 2))
print("Average price in lakhs: ₹", round(average_price / 100000, 2), "lakh")
```

```python
location_summary = df.groupby("location")["price"].mean().sort_values(
    ascending=False
)

display(
    (location_summary / 100000)
    .round(2)
    .rename("Average Price (₹ lakh)")
    .reset_index()
)

location_summary_lakhs = location_summary / 100000

plt.figure(figsize=(8, 5))
plt.bar(location_summary_lakhs.index, location_summary_lakhs.values)
plt.title("Average House Price by Location")
plt.xlabel("Location")
plt.ylabel("Average Price (₹ lakh)")
plt.grid(axis="y")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["price"] / 100000, bins=5, edgecolor="black")
plt.title("House Price Distribution")
plt.xlabel("House Price (₹ lakh)")
plt.ylabel("Number of Houses")
plt.grid(axis="y")
plt.show()
```

```python
df.to_csv("day30_house_price_dashboard_data.csv", index=False)

location_summary.to_frame(
    name="average_price"
).to_csv("day30_location_summary.csv")

print("Dashboard data saved successfully!")
```

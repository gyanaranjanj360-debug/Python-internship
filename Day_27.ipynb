```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.tree import DecisionTreeRegressor

data = {
    "area_sqft": [800, 1000, 1200, 1500, 1800, 2000, 2200, 2500, 900, 1600],
    "bedrooms":  [2, 2, 3, 3, 3, 4, 4, 4, 2, 3],
    "bathrooms": [1, 2, 2, 2, 3, 3, 3, 4, 1, 2],
    "parking":   [0, 1, 1, 1, 1, 2, 2, 2, 0, 1],
    "price":     [2000000, 2500000, 3200000, 4000000, 4800000,
                  5500000, 6200000, 7500000, 2200000, 4200000]
}

df = pd.DataFrame(data)
display(df)
```

```python
X = df[["area_sqft", "bedrooms", "bathrooms", "parking"]]
y = df["price"]

model = DecisionTreeRegressor(max_depth=3, random_state=42)
model.fit(X, y)

print("Model trained successfully!")
```

```python
importance_df = pd.DataFrame({
    "Feature": X.columns,
    "Importance": model.feature_importances_
})

importance_df = importance_df.sort_values(
    by="Importance", ascending=False
)

display(importance_df)
```

```python
plt.figure(figsize=(8, 5))
plt.bar(importance_df["Feature"], importance_df["Importance"])
plt.title("Day 27: House Price Feature Importance")
plt.xlabel("House Features")
plt.ylabel("Importance Score")
plt.xticks(rotation=20)
plt.grid(axis="y")
plt.show()
```

```python
importance_df.to_csv("day27_feature_importance.csv", index=False)
print("Results saved as day27_feature_importance.csv")
```

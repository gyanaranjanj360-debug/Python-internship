```python
import pandas as pd
import matplotlib.pyplot as plt

data = {
    "area_sqft": [800, 1000, 1200, 1500, 1800, 2000, 2200, 2500, 900, 1600],
    "bedrooms":  [2, 2, 3, 3, 3, 4, 4, 4, 2, 3],
    "bathrooms": [1, 2, 2, 2, 3, 3, 3, 4, 1, 2],
    "parking":   [0, 1, 1, 1, 1, 2, 2, 2, 0, 1],
    "price":     [2000000, 2500000, 3200000, 4000000, 4800000,
                  5500000, 6200000, 7500000, 2200000, 4200000]
}

df = pd.DataFrame(data)
display(df)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor

X = df[["area_sqft", "bedrooms", "bathrooms", "parking"]]
y = df["price"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

model = DecisionTreeRegressor(max_depth=3, random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)

print("Model trained and tested successfully!")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Mean Absolute Error (MAE): ₹", round(mae, 2))
print("Mean Squared Error (MSE):", round(mse, 2))
print("R-squared (R²):", round(r2, 3))
```

```python
results = pd.DataFrame({
    "Actual Price": y_test.values,
    "Predicted Price": y_pred.round(2)
})

display(results)

plt.figure(figsize=(8, 5))
plt.plot(results["Actual Price"].values, marker="o", label="Actual")
plt.plot(results["Predicted Price"].values, marker="s", label="Predicted")
plt.title("Day 26: Actual vs Predicted Prices")
plt.xlabel("Test House")
plt.ylabel("Price (₹)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
errors = y_test.values - y_pred

plt.figure(figsize=(8, 5))
plt.bar(range(len(errors)), errors)
plt.axhline(y=0, linestyle="--")
plt.title("Prediction Errors")
plt.xlabel("Test House")
plt.ylabel("Actual - Predicted (₹)")
plt.grid(axis="y")
plt.show()
```
